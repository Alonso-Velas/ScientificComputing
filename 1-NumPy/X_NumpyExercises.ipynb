# NumPy exercises

```python
#import the Numpy package
import numpy as np
```

```python
#Set the random seed
np.random.seed(859)
```

```python
#Create a random array of 100 values between 0 and 1
arr_1 = np.random.rand(100)

#Display the resulting array
arr_1
```

```python
#Report the shape of the resulting array
arr_1.shape
```

```python
#Report the data type of the resulting array
type(arr_1)
```

```python
#Report the mean value of all values in your array
arr_1.mean()
```

```python
#Multiply each value in your array by 10 to get values between 0 and 10, saving the results as "arr_2"
arr_2 = arr_1 * 10
```

```python
#Reshape the arr_2 array to be 25 rows and 4 columns, saving the result as "arr_3"
arr_3 = arr_2.reshape(25,4)
arr_3.shape
```

```python
#Show the 3rd row of the arr_3 array
arr_3[2,]
```

```python
#Show the values in the 2nd column of the arr_3 array
arr_3[:,1]
```

```python
#Show the mean of the values in the 2nd column
print(arr_3[:,1].mean())
```
